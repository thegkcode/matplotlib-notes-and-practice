```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10, 10))

fig.suptitle("Sales Report 2025", fontsize=18)         # title of main figure

ax[0,0].scatter(batters['avg'], batters['strike_rate'], color='red')
ax[0,1].scatter(batters['avg'], batters['runs'], marker='d')
ax[1,0].hist(batters['avg'], color='orange', edgecolor='black')
ax[1,1].hist(batters['runs'], color='green', edgecolor='black', log=True)

ax[0,0].set_title('Avg vs Strike rate')
ax[0,1].set_title('Avg vs Runs')
ax[1,0].set_title('Average - Frequency')
ax[1,1].set_title('Runs - Frequency')

ax[0,0].set_xlabel('Avg..')
ax[0,0].set_ylabel('Avg..')
```

## 3D Surface 

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
import numpy as np

x = np.array([1,2,3])
y = np.array([10,20])

X, Y = np.meshgrid(x, y)

print(X.shape)
print(Y)

# X.shape = (2,3)
# Y.shape = (2,3)
# rows = len(y)
# cols = len(x)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax=plt.subplot(projection='3d')

# ax.plot_surface(xx, yy, z, color='yellow')
# ax.plot_surface(xx, yy, z, cmap='viridis')
p=ax.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(12,8))

ax=plt.subplot(projection='3d')

p=ax.plot_surface(xx, yy, z, cmap='jet')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))
z = xx**2 + yy**2
ax=plt.subplot(projection='3d')

# ax.plot_surface(xx, yy, z, color='yellow')
ax.plot_surface(xx, yy, z, cmap='viridis')
```

## Contour Plots
### -> is a special plot that represent 3D graph in 2D 

```python
fig = plt.figure(figsize=(12,8))
z = xx**2 + yy**2
ax=plt.subplot()

p=ax.contour(xx, yy, z, cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))
z = xx**2 + yy**2
ax=plt.subplot()

p=ax.contourf(xx, yy, z, cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(12,8))

ax=plt.subplot()

p=ax.contourf(xx, yy, z, cmap='jet')
fig.colorbar(p)
```

## Heat Map

```python
df=pd.read_csv(r'IPL_Ball_by_Ball_2008_2022.csv')
df.head()
```

```python
temp_df = df[(df['ballnumber'].isin([1,2,3,4,5,6])) & (df['batsman_run']==6)]
```

```python
pd.pivot_table(temp_df, index='overs', columns='ballnumber', values='batsman_run', aggfunc='count')
```

```python
 grid = pd.crosstab(temp_df['overs'], df['ballnumber'])
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.colorbar()
```
