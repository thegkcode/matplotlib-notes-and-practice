```python
# plt.plot(x, y, color='color name', linestyle=' -- or - or : or :.', linewidth=value, marker=' o or s or d or + or *', label='label_name')
# plt.xlabel("text")
# plt.ylabel("kuchh bhi text")
# plt.legend(loc='upper left / lower right / upper right /  lower left', fontsize=value)    loc,fontsize are Optional
# plt.xlim(strt, end)
# plt.ylim(strt, end)

months=[1,2,3,4]
sales=[1000,1500,1200,2000]

plt.plot(months, sales, color='blue', linestyle='--', linewidth=2, marker='o', label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='gray', linestyle=':', linewidth=1)
# plt.grid(
#     visible=True,      # grid show/hide
#     axis='both',       # 'x', 'y', 'both'
#     color='gray',      # grid color
#     linestyle='--',    # '-', '--', ':', '-.'
#     linewidth=1,       # line thickness
#     alpha=0.5          # transparency (0 to 1)
# )
plt.xticks(months, ['M1', 'M2', 'M3', 'M4'], rotation=45, color='green')    # same -> plt.yticks()
plt.xlim(1,4)
plt.ylim(0,2000)
plt.show()
```

# Bar chart
### Bivariate Analysis , also for univaraite

```python
# Simple bar chart
children = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, children, color='black')
```

```python
# Stacked Bar Chart
plt.bar(df['batsman'], df['2017'], label='2017')
plt.bar(df['batsman'], df['2016'], bottom=df['2017'], label='2016')
plt.bar(df['batsman'], df['2015'], bottom=(df['2016'] + df['2017']), label='2015')
plt.legend()
```

# Pie Chart
### Univariate/Bivariate Analysis
### Categorical vs numerical
### Use case - To find contribution on a standard scale(means 100% , ek category ka kitna %, doosri category ka kitna %)

```python
# Simple data
data = [23,45,100,20,49]
marks = ['English', 'Hindi', 'Science', 'S.Sc', 'Maths']
plt.pie(data, labels=marks)
plt.show()
```

```python
# dataset
df=pd.read_csv(r'gayle-175.csv')
df
```

```python
# percentage
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
# colors
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', colors=['red', 'green', 'blue', 'yellow', 'orange', 'pink'], shadow=True)
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode=[.3,0,0,0,0,.5], shadow=True)
plt.show()
```
